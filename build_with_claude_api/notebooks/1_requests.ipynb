```python
import os 
from dotenv import load_dotenv
load_dotenv()

model = os.environ["CLAUDE_MODEL"]
```

```python
from anthropic import Anthropic
client = Anthropic() 
```

```python
model
```

```python
# Make a Request
message = client.messages.create(
    model = model,
    max_tokens = 1000,
    messages= [
        {
            "role": "user",
            "content": "What is Quantum Computing? Answer in one sentence"
        }
    ]
)
```
